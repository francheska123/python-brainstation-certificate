## Functions

A function is a named block of code that can be called for reuse. Functions are key to creating efficient code as they give us the ability to reuse code and make our code more flexible with the use of parameters. Before we discuss the use of parameters, let's look at how we can write a function.

The structure of a function has four main parts:
- The definition keyword: `def`
- The information you want to pass in: parameters
- The code you want to execute
- The `return` keyword

After we create a function, it can be executed by "calling" or "invoking" the function name. To do this, you simply write the name of the function followed by a set of parentheses `()`. If the function requires parameters, you place them inside the parentheses separating each with a comma.

````python
# Creating a function
def formatToDogYears(name, age):
    formattedAge = name + " is " + str(age * 7) + ' years old in dog years.'
    return formattedAge

# Invoking a function
formatToDogYears('Spike', 5)
````

````python
# Define a function to return the full name by passing the name components

def getpersonfullname(first_name, middle_name, last_name):
    full_name = first_name + " " + middle_name + " " + last_name
    return full_name
````

````python
getpersonfullname('Alex', 'R', 'Donald')
````

### Anatomy of a Function



### Parameters vs Arguments
A `parameter` refers to a variable that can be found within our function definition code. Parameters give us the ability to pass information into a function where it is essential to produce a particular result. Think of a parameter as a placeholder, it contains no actual value. Once a parameter contains a value, it is known as an `argument`. The term parameter and argument is sometimes used interchangeably but is useful to know that a parameter only becomes an argument once a value is passed to the function.

If we refer to the function example above, "name" is the parameter while "Spike" is the argument. When the function is invoked, "Spike" replaces the name parameter on line two.

### The `return` Statement
The caller passes information to a function via parameters / arguments in order to run some computation on those values. Once that computation is complete, the function should pass information back to the caller. The `return` statement is used to pass information "out" of a function or back to the caller. To capture that returned value, a function invocation is assigned to a variable. If the `return` statement is **not** used, the function will return a default value of `None`. To correctly return a value from a function, the value must be on the same line, directly after the return statement.

````python
# Correctly returns the passed value multiplied by five back to the caller
def multipleOfFive(num):
    return num * 5

correct_use = multipleOfFive(5)

# No return statment - returns a value of None
def multipleOfFive(num):
    num * 5

no_return_statement = multipleOfFive(5)

# Incorrect placement of the return statement - returns a value of None
def multipleOfFive(num):
    num * 5
    return

wrong_return_placement = multipleOfFive(5)

print(correct_use)
print(no_return_statement)
print(wrong_return_placement)
````

### Best Practices

Python functions are very flexible. You can define a function that can do pretty much anything you can think of. However, there are some best practice guidelines which are worth following:

1. Functions should always return the same output when given the same input.

2. Functions should only affect variables that have been passed into the function via arguments.
    - You may see this referred to as 'no side effects'. This helps prevent any unwanted side effects and makes your code easier to maintain and test.


3. Break down functions into individual units of work.
    
    - This is also known as the 'single responsibility principle'. Large functions, which carry out multiple different tasks, become unwieldy, hard to maintain, hard to decipher, and hard to test. Breaking complex functions into smaller units will provide many benefits down the line.

Let's take a closer look at tips 1 and 2.

#### Pandas: Cleaning

Data Cleaning or Data Cleansing is done to detect and correct or remove corrupted data. Data cleaning can be done using pure Python but Pandas makes this task much easier.

Some common cleaning techniques for removing NaN or duplicated elements are as followed:


- Row with any duplicates: `dataframe.drop_duplicates()`
- Rows with duplicated elements in specific columns: `dataframe.drop_duplicates(subset=["name", "age"])`
- Rows with any missing data: `dataframe.dropna()`
- Rows with all elements missing: `dataframe.dropna(how="all")`
- Rows with missing elements for specific columns: `dataframe.dropna(subset=["name", "age"])`
- Columns with any missing element: `dataframe.dropna(axis="columns")`

#### Pandas: Aggregation

An _aggregating function_ is a function that takes a bunch of values and returns a single number. Some examples include `sum`, `mean`, `len`, etc. An _aggregation_ on a grouped dataframe applies an aggregating function to each group, and returns a dataframe with a single row per group.

Aggregation can be done in Pandas using `.group_by()` function, which is then associated with one of the aggregation functions such as `sum`, `mean`, or `len`.

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Program": ["Software Engineering", "Data Science", "UI/UX", "Product Management",
                 "Digital Marketing", "Software Engineering", "Data Science", "UI/UX",
                 "Product Management", "Digital Marketing"],
    "Market Demand": [70, 100, 80, 70, 150, 160, 140, 200, 90, 130],
    "Hiring": [30, 20, 30, 22, 27, 29, 18, 25, 30, 40]},
    index = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
)

df
````

````python
#gets the first five rows
df.head()
````

#### Q1:

Define a list of words and create a function to find the shortest word.

````python
# Write code here...

#Define list of words

# Step 1: Define the list of words
words = ["Desk", "Lamp", "Monitor", "Computer", "Phone", "Notepad", "Pen", "Pencil"]

# Step 2: Create a function to find the shortest word
def find_shortest_word(word_list):
    # Initialize the shortest word as the first word in the list
    shortest_word = word_list[0]
    
    # Iterate over the word list
    for word in word_list:
        # Compare the lengths of words and update the shortest word if necessary
        if len(word) < len(shortest_word):
            shortest_word = word
    
    return shortest_word

# Step 3: Call the function and print the result
shortest = find_shortest_word(words)
print(f"The shortest word is: {shortest}")
````

#### Q3:

Create a function that determines whether a number is Oddish or Evenish. A number is Oddish if the sum of all of its digits is odd, and a number is Evenish if the sum of all of its digits is even. If a function is Oddish, return "Oddish". Otherwise, return "Evenish".


````python
# Write code here...
````

````python
def oddish_or_evenish(n):
    # Convert the number to a string, then sum the digits
    sum_of_digits = sum(int(digit) for digit in str(n))
    
    # Check if the sum is odd or even
    if sum_of_digits % 2 == 0:
        return "Evenish"
    else:
        return "Oddish"

# Example usage
print(oddish_or_evenish(43))  # Output: "Oddish" (4 + 3 = 7)
print(oddish_or_evenish(22))  # Output: "Evenish" (2 + 2 = 4)
````

You are given a 2D list. The task is to find the sum of minimum values from each row:
Example:
    
`[[1, 2, 3, 4, 5],
 [5, 6, 7, 8, 9],
 [20, 21, 34, 56, 100]]`

The function should return 26 (1+5+20=26)

````python
# Write code here...
````

````python
def sum_of_minimums(matrix):
    return sum(min(row) for row in matrix)

# Example usage
matrix = [[1, 2, 3, 4, 5], [5, 6, 7, 8, 9], [20, 21, 34, 56, 100]]
print(sum_of_minimums(matrix))  # Output: 26 (1 + 5 + 20 = 26)
````
